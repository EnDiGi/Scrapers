This program get news articles from various sites and displays them

Library import

```python
import requests
from bs4 import BeautifulSoup
```

Scraping from www.focus.it

```python

url = "https://www.focus.it"
hiddenTopics = ["Focus Talks"]

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

headings = soup.find_all(["h2", "h3"])

contents = [list(h.children)[3].text for h in headings if list(h.children)[1].text not in hiddenTopics]

contents.sort(key = lambda x: x[0])

focusOutput = "FOCUS\n"

for content in contents:
  focusOutput += content + "\n"
```

Scraping from www.geopop.it

```python

geopopOutput = "GEOPOP\n"

url = "https://www.geopop.it"
hiddenTopics = ["Focus Talks"]

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

headings = soup.select("a.crd__title")

for a in headings:
  geopopOutput += a.text + "\n"
```

The output is a list of all the articles

```python
print(focusOutput + "\n" + geopopOutput)
```
