This program get news articles from www.focus.it and displays them

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.focus.it"
hiddenTopics = ["Focus Talks"]

response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")

headings = soup.find_all(["h2", "h3"])

contents = [(list(h.children)[1].text, list(h.children)[3].text) for h in headings if list(h.children)[1].text not in hiddenTopics]

contents.sort(key = lambda x: x[0])

message = ""

message += ("ARGOMENTO".ljust(15) + "ARTICOLO\n")
message += ("-" * 32 + "\n")


for content in contents:
  message += (content[0].ljust(15) + content[1] + "\n")

print(message)
```

The output is a table-ish disposition of topics and the respective article
